# Regional Comparison

**All numeric data here are synthetic and are not actual WHO measurements.**

Unweighted averages of selected countries are not official regional aggregates.

```python
import sys
from pathlib import Path
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
import pandas as pd
from healthpulse.ingestion.public_dataset_loader import make_demo_data
df = pd.DataFrame(make_demo_data())
print('Synthetic demo rows:', len(df))
```

```python
from healthpulse.analytics.regional_comparison import compare_regions
subset=df[df.indicator=='immunization']
regional = compare_regions(subset,year=2024)
display(pd.DataFrame(regional))
```

```python
import matplotlib.pyplot as plt
pd.DataFrame(regional).plot.bar(x='region',y='value',legend=False,title='DEMO: Unweighted immunization comparison')
plt.ylabel('coverage %'); plt.tight_layout(); plt.show()
```
