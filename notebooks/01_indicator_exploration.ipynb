# Indicator Exploration

**All numeric data here are synthetic and are not actual WHO measurements.**

Evaluate coverage, units and trends, treating demo values as synthetic.

```python
import sys
from pathlib import Path
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
import pandas as pd
from healthpulse.ingestion.public_dataset_loader import make_demo_data
df = pd.DataFrame(make_demo_data())
print('Synthetic demo rows:', len(df))
```

```python
display(df.head())
print(df.groupby('indicator').agg(countries=('country','nunique'),years=('year','nunique'),observations=('value','size')))
```

```python
import matplotlib.pyplot as plt
pivot = df[df.indicator == 'life_expectancy'].pivot(index='year',columns='country',values='value')
pivot[['IND','USA','JPN']].plot(figsize=(11,4),title='DEMO: Synthetic life expectancy trajectories')
plt.ylabel('years'); plt.show()
```
