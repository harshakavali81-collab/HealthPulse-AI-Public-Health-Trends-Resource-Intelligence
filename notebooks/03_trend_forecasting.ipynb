# Trend Forecasting and Evaluation

**All numeric data here are synthetic and are not actual WHO measurements.**

Chronological train/test separation is used to minimize leakage. No epidemiological validation is claimed.

```python
import sys
from pathlib import Path
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
import pandas as pd
from healthpulse.ingestion.public_dataset_loader import make_demo_data
df = pd.DataFrame(make_demo_data())
print('Synthetic demo rows:', len(df))
```

```python
from healthpulse.models.forecast import forecast
series=df[(df.country=='IND') & (df.indicator=='life_expectancy')].to_dict(orient='records')
result=forecast(series,'life_expectancy',horizon=3)
print('Holdout MAE:',result['holdout_mae'],'Naive MAE:',result['baseline_mae'])
display(pd.DataFrame(result['forecast']))
```

```python
import matplotlib.pyplot as plt
observed=pd.DataFrame(series)[['year','value']]
future=pd.DataFrame(result['forecast'])
plt.plot(observed.year,observed.value,label='Synthetic observed')
plt.plot([observed.year.iloc[-1]]+future.year.tolist(),[observed.value.iloc[-1]]+future.value.tolist(),'--',label='Illustrative trend')
plt.title('DEMO: Ridge trend projection, not a health prediction'); plt.legend(); plt.show()
```
